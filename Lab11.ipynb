```python
# ============================================================
#  Lab 13 - Natural Language Processing (NLP)
#  Dataset: TCGA GBM/LGG Mutations
#  Target: Predict Glioma Grade from text features
#         (Primary_Diagnosis, Race, Gender)
# ============================================================

# ── 1. Import Libraries ─────────────────────────────────────────
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import string
import warnings
warnings.filterwarnings('ignore')

# NLP libraries
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.probability import FreqDist
from collections import Counter

# Sklearn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import (classification_report, confusion_matrix,
                              accuracy_score, roc_auc_score, roc_curve)

# Download NLTK data
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')

print("All libraries imported successfully!")

# ── 2. Load Dataset ─────────────────────────────────────────────
df = pd.read_csv("TCGA_GBM_LGG_Mutations_all.csv")
print(f"Dataset shape: {df.shape}")
print(f"\nColumn names:\n{list(df.columns)}")
df.head(10)

# ── 3. EDA on Text Columns ──────────────────────────────────────
print("\nText columns we will use for NLP:")
text_cols = ['Primary_Diagnosis', 'Race', 'Gender']
for col in text_cols:
    print(f"\n── {col} ──")
    print(df[col].value_counts())

# Grade distribution
print("\nGrade distribution (target):")
print(df['Grade'].value_counts())

plt.figure(figsize=(6, 4))
sns.countplot(x='Grade', data=df, palette='Set2')
plt.title('Grade Distribution (LGG vs GBM)', fontsize=14, fontweight='bold')
plt.xlabel('Grade')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

# Primary Diagnosis distribution
plt.figure(figsize=(10, 5))
df['Primary_Diagnosis'].value_counts().plot(kind='bar', color='steelblue')
plt.title('Primary Diagnosis Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Diagnosis')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Primary Diagnosis by Grade
plt.figure(figsize=(12, 5))
pd.crosstab(df['Primary_Diagnosis'], df['Grade']).plot(
    kind='bar', figsize=(12, 5), colormap='Set2')
plt.title('Primary Diagnosis by Grade', fontsize=14, fontweight='bold')
plt.xlabel('Primary Diagnosis')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Grade')
plt.tight_layout()
plt.show()

# ── 4. Create Text Feature ──────────────────────────────────────
# Combine all text columns into one document per sample
df['text'] = (df['Primary_Diagnosis'].astype(str) + ' ' +
              df['Race'].astype(str) + ' ' +
              df['Gender'].astype(str))

print("\nSample text documents:")
for i in range(5):
    print(f"  [{i}] Grade={df['Grade'].iloc[i]} | Text: {df['text'].iloc[i]}")

# Target: binary (LGG=0, GBM=1)
df['target'] = (df['Grade'] == 'GBM').astype(int)
print(f"\nTarget distribution:\n{df['target'].value_counts()}")

# ── 5. Text Preprocessing ───────────────────────────────────────
stop_words = set(stopwords.words('english'))
stemmer    = PorterStemmer()
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    # Lowercase
    text = text.lower()
    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    # Remove numbers
    text = re.sub(r'\d+', '', text)
    # Tokenize
    tokens = word_tokenize(text)
    # Remove stopwords
    tokens = [t for t in tokens if t not in stop_words]
    # Remove short tokens
    tokens = [t for t in tokens if len(t) > 2]
    # Lemmatize
    tokens = [lemmatizer.lemmatize(t) for t in tokens]
    return ' '.join(tokens)

df['text_clean'] = df['text'].apply(preprocess_text)

print("\nBefore vs After Preprocessing:")
for i in range(3):
    print(f"\nOriginal : {df['text'].iloc[i]}")
    print(f"Cleaned  : {df['text_clean'].iloc[i]}")

# ── 6. Tokenization & Word Frequency Analysis ───────────────────
all_words = ' '.join(df['text_clean']).split()
word_freq = Counter(all_words)

print(f"\nTotal unique words: {len(word_freq)}")
print(f"\nTop 20 most common words:")
for word, count in word_freq.most_common(20):
    print(f"  {word:<25} {count}")

# Word frequency plot
top_words = pd.Series(dict(word_freq.most_common(20)))
plt.figure(figsize=(12, 5))
top_words.plot(kind='bar', color='coral')
plt.title('Top 20 Most Frequent Words', fontsize=14, fontweight='bold')
plt.xlabel('Word')
plt.ylabel('Frequency')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# ── 7. Stemming vs Lemmatization Demo ──────────────────────────
sample_words = ['oligodendroglioma', 'glioblastoma', 'astrocytoma',
                'diagnosis', 'anaplastic', 'mutation', 'running',
                'better', 'studies']

print("\nStemming vs Lemmatization:")
print(f"{'Word':<20} {'Stemmed':<20} {'Lemmatized':<20}")
print("-" * 60)
for w in sample_words:
    print(f"{w:<20} {stemmer.stem(w):<20} {lemmatizer.lemmatize(w):<20}")

# ── 8. Word-level analysis by Grade ────────────────────────────
lgG_text = ' '.join(df[df['target'] == 0]['text_clean']).split()
gbm_text = ' '.join(df[df['target'] == 1]['text_clean']).split()

lgg_freq = Counter(lgG_text)
gbm_freq = Counter(gbm_text)

print("\nTop 10 words in LGG samples:")
for w, c in lgg_freq.most_common(10):
    print(f"  {w:<25} {c}")

print("\nTop 10 words in GBM samples:")
for w, c in gbm_freq.most_common(10):
    print(f"  {w:<25} {c}")

# Side by side bar chart
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
lgg_top = pd.Series(dict(lgg_freq.most_common(10)))
gbm_top = pd.Series(dict(gbm_freq.most_common(10)))

lgg_top.plot(kind='barh', ax=axes[0], color='steelblue')
axes[0].set_title('Top 10 Words - LGG', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Frequency')

gbm_top.plot(kind='barh', ax=axes[1], color='tomato')
axes[1].set_title('Top 10 Words - GBM', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Frequency')

plt.tight_layout()
plt.show()

# ── 9. Feature Extraction ───────────────────────────────────────
X = df['text_clean']
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y)

print(f"\nTraining samples: {len(X_train)}")
print(f"Testing samples:  {len(X_test)}")

# ── Method 1: Bag of Words (CountVectorizer) ────────────────────
print("\n── Bag of Words (CountVectorizer) ──")
cv = CountVectorizer(max_features=500, ngram_range=(1, 2))
X_train_cv = cv.fit_transform(X_train)
X_test_cv  = cv.transform(X_test)

print(f"Vocabulary size:        {len(cv.vocabulary_)}")
print(f"Train matrix shape:     {X_train_cv.shape}")
print(f"Test matrix shape:      {X_test_cv.shape}")
print(f"\nSample vocabulary (first 20):")
print(list(cv.vocabulary_.keys())[:20])

# ── Method 2: TF-IDF ────────────────────────────────────────────
print("\n── TF-IDF Vectorizer ──")
tfidf = TfidfVectorizer(max_features=500, ngram_range=(1, 2))
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf  = tfidf.transform(X_test)

print(f"Vocabulary size:        {len(tfidf.vocabulary_)}")
print(f"Train matrix shape:     {X_train_tfidf.shape}")
print(f"Test matrix shape:      {X_test_tfidf.shape}")

# Top TF-IDF terms
tfidf_scores = pd.Series(
    tfidf.idf_, index=tfidf.get_feature_names_out()
).sort_values(ascending=True)

print(f"\nTop 15 highest TF-IDF terms (most informative):")
print(tfidf_scores.head(15).to_string())

plt.figure(figsize=(10, 5))
tfidf_scores.head(15).plot(kind='barh', color='teal')
plt.title('Top 15 TF-IDF Terms', fontsize=14, fontweight='bold')
plt.xlabel('IDF Score (lower = more informative)')
plt.tight_layout()
plt.show()

# ── 10. Model Training & Evaluation ────────────────────────────

def evaluate_model(model, X_tr, X_te, y_tr, y_te,
                   model_name, vectorizer_name):
    model.fit(X_tr, y_tr)
    y_pred = model.predict(X_te)

    # For ROC-AUC
    if hasattr(model, 'predict_proba'):
        y_proba = model.predict_proba(X_te)[:, 1]
    elif hasattr(model, 'decision_function'):
        y_proba = model.decision_function(X_te)
    else:
        y_proba = y_pred

    acc = accuracy_score(y_te, y_pred)
    try:
        auc = roc_auc_score(y_te, y_proba)
    except:
        auc = 0.0

    print(f"\n{'='*55}")
    print(f"  {model_name} + {vectorizer_name}")
    print(f"{'='*55}")
    print(f"  Accuracy:  {acc:.4f} ({acc*100:.2f}%)")
    print(f"  ROC-AUC:   {auc:.4f}")
    print(f"\n  Confusion Matrix:")
    cm = confusion_matrix(y_te, y_pred)
    print(f"  {cm}")
    print(f"\n  Classification Report:")
    print(classification_report(y_te, y_pred,
          target_names=['LGG', 'GBM']))

    return model, y_pred, y_proba, acc, auc

results = {}

# ── Naive Bayes + BoW ──────────────────────────────────────────
nb_bow, nb_bow_pred, nb_bow_proba, nb_bow_acc, nb_bow_auc = evaluate_model(
    MultinomialNB(), X_train_cv, X_test_cv, y_train, y_test,
    'Naive Bayes', 'Bag of Words')
results['NB + BoW'] = {'acc': nb_bow_acc, 'auc': nb_bow_auc}

# Confusion matrix heatmap
plt.figure(figsize=(5, 4))
sns.heatmap(confusion_matrix(y_test, nb_bow_pred), annot=True, fmt='d',
            cmap='Blues', xticklabels=['LGG','GBM'], yticklabels=['LGG','GBM'])
plt.title('Confusion Matrix\nNaive Bayes + BoW', fontweight='bold')
plt.ylabel('Actual'); plt.xlabel('Predicted')
plt.tight_layout(); plt.show()

# ── Naive Bayes + TF-IDF ───────────────────────────────────────
nb_tfidf, nb_tfidf_pred, nb_tfidf_proba, nb_tfidf_acc, nb_tfidf_auc = evaluate_model(
    MultinomialNB(), X_train_tfidf, X_test_tfidf, y_train, y_test,
    'Naive Bayes', 'TF-IDF')
results['NB + TF-IDF'] = {'acc': nb_tfidf_acc, 'auc': nb_tfidf_auc}

plt.figure(figsize=(5, 4))
sns.heatmap(confusion_matrix(y_test, nb_tfidf_pred), annot=True, fmt='d',
            cmap='Blues', xticklabels=['LGG','GBM'], yticklabels=['LGG','GBM'])
plt.title('Confusion Matrix\nNaive Bayes + TF-IDF', fontweight='bold')
plt.ylabel('Actual'); plt.xlabel('Predicted')
plt.tight_layout(); plt.show()

# ── Logistic Regression + BoW ──────────────────────────────────
lr_bow, lr_bow_pred, lr_bow_proba, lr_bow_acc, lr_bow_auc = evaluate_model(
    LogisticRegression(max_iter=1000, random_state=42),
    X_train_cv, X_test_cv, y_train, y_test,
    'Logistic Regression', 'Bag of Words')
results['LR + BoW'] = {'acc': lr_bow_acc, 'auc': lr_bow_auc}

plt.figure(figsize=(5, 4))
sns.heatmap(confusion_matrix(y_test, lr_bow_pred), annot=True, fmt='d',
            cmap='Oranges', xticklabels=['LGG','GBM'], yticklabels=['LGG','GBM'])
plt.title('Confusion Matrix\nLogistic Regression + BoW', fontweight='bold')
plt.ylabel('Actual'); plt.xlabel('Predicted')
plt.tight_layout(); plt.show()

# ── Logistic Regression + TF-IDF ──────────────────────────────
lr_tfidf, lr_tfidf_pred, lr_tfidf_proba, lr_tfidf_acc, lr_tfidf_auc = evaluate_model(
    LogisticRegression(max_iter=1000, random_state=42),
    X_train_tfidf, X_test_tfidf, y_train, y_test,
    'Logistic Regression', 'TF-IDF')
results['LR + TF-IDF'] = {'acc': lr_tfidf_acc, 'auc': lr_tfidf_auc}

plt.figure(figsize=(5, 4))
sns.heatmap(confusion_matrix(y_test, lr_tfidf_pred), annot=True, fmt='d',
            cmap='Oranges', xticklabels=['LGG','GBM'], yticklabels=['LGG','GBM'])
plt.title('Confusion Matrix\nLogistic Regression + TF-IDF', fontweight='bold')
plt.ylabel('Actual'); plt.xlabel('Predicted')
plt.tight_layout(); plt.show()

# ── SVM + BoW ──────────────────────────────────────────────────
svm_bow, svm_bow_pred, svm_bow_proba, svm_bow_acc, svm_bow_auc = evaluate_model(
    LinearSVC(random_state=42, max_iter=2000),
    X_train_cv, X_test_cv, y_train, y_test,
    'SVM (LinearSVC)', 'Bag of Words')
results['SVM + BoW'] = {'acc': svm_bow_acc, 'auc': svm_bow_auc}

plt.figure(figsize=(5, 4))
sns.heatmap(confusion_matrix(y_test, svm_bow_pred), annot=True, fmt='d',
            cmap='Greens', xticklabels=['LGG','GBM'], yticklabels=['LGG','GBM'])
plt.title('Confusion Matrix\nSVM + BoW', fontweight='bold')
plt.ylabel('Actual'); plt.xlabel('Predicted')
plt.tight_layout(); plt.show()

# ── SVM + TF-IDF ───────────────────────────────────────────────
svm_tfidf, svm_tfidf_pred, svm_tfidf_proba, svm_tfidf_acc, svm_tfidf_auc = evaluate_model(
    LinearSVC(random_state=42, max_iter=2000),
    X_train_tfidf, X_test_tfidf, y_train, y_test,
    'SVM (LinearSVC)', 'TF-IDF')
results['SVM + TF-IDF'] = {'acc': svm_tfidf_acc, 'auc': svm_tfidf_auc}

plt.figure(figsize=(5, 4))
sns.heatmap(confusion_matrix(y_test, svm_tfidf_pred), annot=True, fmt='d',
            cmap='Greens', xticklabels=['LGG','GBM'], yticklabels=['LGG','GBM'])
plt.title('Confusion Matrix\nSVM + TF-IDF', fontweight='bold')
plt.ylabel('Actual'); plt.xlabel('Predicted')
plt.tight_layout(); plt.show()

# ── 11. ROC Curve Comparison ────────────────────────────────────
plt.figure(figsize=(10, 7))

# NB + BoW
fpr, tpr, _ = roc_curve(y_test, nb_bow_proba)
plt.plot(fpr, tpr, lw=2, label=f'NB + BoW       (AUC = {nb_bow_auc:.4f})')

# NB + TF-IDF
fpr, tpr, _ = roc_curve(y_test, nb_tfidf_proba)
plt.plot(fpr, tpr, lw=2, label=f'NB + TF-IDF    (AUC = {nb_tfidf_auc:.4f})')

# LR + BoW
fpr, tpr, _ = roc_curve(y_test, lr_bow_proba)
plt.plot(fpr, tpr, lw=2, label=f'LR + BoW       (AUC = {lr_bow_auc:.4f})')

# LR + TF-IDF
fpr, tpr, _ = roc_curve(y_test, lr_tfidf_proba)
plt.plot(fpr, tpr, lw=2, label=f'LR + TF-IDF    (AUC = {lr_tfidf_auc:.4f})')

plt.plot([0, 1], [0, 1], 'k--', lw=1.5, label='Random Classifier')
plt.xlabel('False Positive Rate', fontsize=12)
plt.ylabel('True Positive Rate', fontsize=12)
plt.title('ROC Curve Comparison - All NLP Models', fontsize=14, fontweight='bold')
plt.legend(loc='lower right', fontsize=10)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# ── 12. Model Comparison Summary ───────────────────────────────
print("\n── Model Comparison Summary ──")
comparison_df = pd.DataFrame({
    'Model': list(results.keys()),
    'Accuracy': [v['acc'] for v in results.values()],
    'ROC-AUC':  [v['auc'] for v in results.values()]
}).sort_values('Accuracy', ascending=False)

print(comparison_df.to_string(index=False))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Accuracy
axes[0].barh(comparison_df['Model'], comparison_df['Accuracy'],
             color='steelblue')
axes[0].set_title('Model Accuracy Comparison', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Accuracy')
axes[0].set_xlim(0, 1)
for i, v in enumerate(comparison_df['Accuracy']):
    axes[0].text(v + 0.01, i, f'{v:.4f}', va='center', fontsize=10)

# ROC-AUC
axes[1].barh(comparison_df['Model'], comparison_df['ROC-AUC'],
             color='coral')
axes[1].set_title('Model ROC-AUC Comparison', fontsize=13, fontweight='bold')
axes[1].set_xlabel('ROC-AUC Score')
axes[1].set_xlim(0, 1)
for i, v in enumerate(comparison_df['ROC-AUC']):
    axes[1].text(v + 0.01, i, f'{v:.4f}', va='center', fontsize=10)

plt.tight_layout()
plt.show()

# ── 13. N-gram Analysis ─────────────────────────────────────────
print("\n── N-gram Analysis ──")

# Unigrams
cv_uni = CountVectorizer(max_features=10, ngram_range=(1, 1))
cv_uni.fit(df['text_clean'])
print("Top 10 Unigrams:", list(cv_uni.vocabulary_.keys()))

# Bigrams
cv_bi = CountVectorizer(max_features=10, ngram_range=(2, 2))
cv_bi.fit(df['text_clean'])
print("Top 10 Bigrams: ", list(cv_bi.vocabulary_.keys()))

# Trigrams
cv_tri = CountVectorizer(max_features=10, ngram_range=(3, 3))
cv_tri.fit(df['text_clean'])
print("Top 10 Trigrams:", list(cv_tri.vocabulary_.keys()))

# Bigram frequency bar chart
bigram_vec = CountVectorizer(max_features=15, ngram_range=(2, 2))
X_bigram = bigram_vec.fit_transform(df['text_clean'])
bigram_counts = pd.Series(
    X_bigram.toarray().sum(axis=0),
    index=bigram_vec.get_feature_names_out()
).sort_values(ascending=False)

plt.figure(figsize=(12, 5))
bigram_counts.plot(kind='bar', color='mediumslateblue')
plt.title('Top 15 Bigrams in Dataset', fontsize=14, fontweight='bold')
plt.xlabel('Bigram')
plt.ylabel('Frequency')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# ── 14. Feature Importance from Logistic Regression ────────────
print("\n── Top Features from Logistic Regression (TF-IDF) ──")
feature_names = tfidf.get_feature_names_out()
coef = lr_tfidf.coef_[0]
top_pos_idx = coef.argsort()[-15:][::-1]
top_neg_idx = coef.argsort()[:15]

top_features = pd.DataFrame({
    'Feature':     list(feature_names[top_pos_idx]) + list(feature_names[top_neg_idx]),
    'Coefficient': list(coef[top_pos_idx]) + list(coef[top_neg_idx]),
    'Class':       ['GBM'] * 15 + ['LGG'] * 15
})

print("\nTop 15 features predicting GBM:")
print(top_features[top_features['Class'] == 'GBM'][['Feature','Coefficient']].to_string(index=False))

print("\nTop 15 features predicting LGG:")
print(top_features[top_features['Class'] == 'LGG'][['Feature','Coefficient']].to_string(index=False))

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

gbm_feat = top_features[top_features['Class'] == 'GBM'].sort_values('Coefficient')
lgg_feat = top_features[top_features['Class'] == 'LGG'].sort_values('Coefficient', ascending=False)

axes[0].barh(gbm_feat['Feature'], gbm_feat['Coefficient'], color='tomato')
axes[0].set_title('Top Features → GBM', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Coefficient')

axes[1].barh(lgg_feat['Feature'], abs(lgg_feat['Coefficient']), color='steelblue')
axes[1].set_title('Top Features → LGG', fontsize=13, fontweight='bold')
axes[1].set_xlabel('|Coefficient|')

plt.suptitle('Logistic Regression Feature Importance (TF-IDF)',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# ── 15. Predict on New Text Samples ────────────────────────────
print("\n── Predict on New Samples ──")

new_samples = [
    "Glioblastoma white male",
    "Oligodendroglioma NOS white female",
    "Astrocytoma anaplastic white male",
    "Mixed glioma white male",
    "Oligodendroglioma anaplastic white female"
]

new_clean = [preprocess_text(s) for s in new_samples]
new_tfidf = tfidf.transform(new_clean)
new_preds = lr_tfidf.predict(new_tfidf)
new_proba = lr_tfidf.predict_proba(new_tfidf)[:, 1]

print(f"\n{'Input Text':<45} {'Predicted':<12} {'GBM Prob':<10}")
print("-" * 70)
for text, pred, prob in zip(new_samples, new_preds, new_proba):
    label = 'GBM' if pred == 1 else 'LGG'
    print(f"{text:<45} {label:<12} {prob:.4f}")

# ── 16. Final Summary ───────────────────────────────────────────
best_model = comparison_df.iloc[0]
print("\n" + "="*55)
print("  NLP PIPELINE SUMMARY")
print("="*55)
print(f"  Text source:       Primary_Diagnosis + Race + Gender")
print(f"  Preprocessing:     Lowercase, punctuation removal,")
print(f"                     stopword removal, lemmatization")
print(f"  Vectorizers:       Bag of Words, TF-IDF (bigrams)")
print(f"  Models tested:     Naive Bayes, Logistic Regression, SVM")
print(f"  Best model:        {best_model['Model']}")
print(f"  Best Accuracy:     {best_model['Accuracy']:.4f}")
print(f"  Best ROC-AUC:      {best_model['ROC-AUC']:.4f}")
print("="*55)
```
